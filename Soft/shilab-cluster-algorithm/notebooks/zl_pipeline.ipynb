```python
import os
import shutil
import pandas as pd
from tqdm import tqdm  # 这是一个进度条库，可选，建议安装: pip install tqdm

def copy_cluster_images(task_list):
    """
    根据CSV中的聚类结果，将图片复制到对应的簇文件夹中。
    
    Args:
        task_list (list): 包含任务字典的列表，每个字典需包含:
                          'csv_path', 'select_cluster', 'save_dir'
    """
    print(f"🚀 开始处理 {len(task_list)} 个任务...\n")

    for i, task in enumerate(task_list):
        csv_path = task['csv_path']
        target_clusters = task['select_cluster'] # 例如: (0, 2, 5)
        base_save_dir = task['save_dir']
        
        print(f"📂 [任务 {i+1}] 正在读取: {os.path.basename(csv_path)}")
        
        # 1. 读取 CSV 文件
        try:
            # header=0 表示第一行是标题
            df = pd.read_csv(csv_path, header=0)
        except Exception as e:
            print(f"❌ 读取CSV失败: {csv_path}, 错误: {e}")
            continue

        # 2. 筛选符合条件的行
        # 假设第一列是标签(Cluster)，第二列是路径(Path)
        # 使用 iloc 按位置索引，比按列名更通用
        # 确保第一列数据类型与 target_clusters 中的类型一致（通常是int）
        
        # 过滤逻辑：判断第一列的值是否在 target_clusters 元组中
        filtered_df = df[df.iloc[:, 0].isin(target_clusters)]

        if filtered_df.empty:
            print(f"⚠️  在 {csv_path} 中未找到属于簇 {target_clusters} 的数据。")
            continue

        print(f"   -> 选中 {len(filtered_df)} 张图片，准备复制...")

        # 3. 遍历筛选后的数据并复制
        # 使用 tqdm 显示进度条 (如果没安装 tqdm，可以直接用 filtered_df.iterrows())
        for index, row in tqdm(filtered_df.iterrows(), total=len(filtered_df), unit="img"):
            cluster_id = row.iloc[0]  # 第一列：簇号
            src_image_path = row.iloc[1]  # 第二列：图片路径
            
            # 4. 构建目标文件夹路径: save_dir/Cx
            # int(cluster_id) 确保簇号是整数，去掉可能的浮点数后缀
            cluster_folder_name = f"C{int(cluster_id)}"
            target_dir = os.path.join(base_save_dir, cluster_folder_name)
            
            # 如果文件夹不存在，则创建
            os.makedirs(target_dir, exist_ok=True)
            
            # 5. 构建目标文件路径
            # 获取原始文件名 (例如: /path/to/img.jpg -> img.jpg)
            file_name = os.path.basename(src_image_path)
            dst_image_path = os.path.join(target_dir, file_name)
            
            # 6. 执行复制
            try:
                # shutil.copy2 会保留文件的元数据（如创建时间）
                shutil.copy2(src_image_path, dst_image_path)
            except FileNotFoundError:
                print(f"\n❌ 文件未找到 (跳过): {src_image_path}")
            except Exception as e:
                print(f"\n❌ 复制出错: {src_image_path} -> {e}")

    print("\n✅ 所有任务处理完成！")

# ==========================================
# 使用示例
# ==========================================

if __name__ == "__main__":
    # 定义你的任务列表
    tasks = [
        # {
        #     'csv_path': r'I:\Esemble_Learning\model1\result\p6_GoodSampleForPresentation\validation_cohort1\C202400803\cluster_detail.csv',
        #     'select_cluster': [0],       # 只提取 0号 和 3号 簇
        #     'save_dir': r'I:\Esemble_Learning\model1\result\p6_GoodSampleForPresentation\validation_cohort1\C202400803\Filtered_cells'
        # },
        # {
        #     'csv_path': r'I:\Esemble_Learning\model1\result\p6_GoodSampleForPresentation\validation_cohort1\C202400048\cluster_detail.csv',
        #     'select_cluster': [1],    # 提取 1, 2, 5号 簇
        #     'save_dir': r'I:\Esemble_Learning\model1\result\p6_GoodSampleForPresentation\validation_cohort1\C202400048\Filtered_cells'
        # }
        {
            'csv_path': r'I:\Esemble_Learning\model1\result\p6_GoodSampleForPresentation\external_cohort1\C202405439\cluster_detail.csv',
            'select_cluster': [0,2,3],       # 只提取 0号 和 3号 簇
            'save_dir': r'I:\Esemble_Learning\model1\result\p6_GoodSampleForPresentation\external_cohort1\C202405439\Filtered_cells'
        },
        {
            'csv_path': r'I:\Esemble_Learning\model1\result\p6_GoodSampleForPresentation\external_cohort1\YC202427807\cluster_detail.csv',
            'select_cluster': [0,1,3],    # 提取 1, 2, 5号 簇
            'save_dir': r'I:\Esemble_Learning\model1\result\p6_GoodSampleForPresentation\external_cohort1\YC202427807\Filtered_cells'
        }
    ]

    # 运行函数
    # 注意：运行前请确保安装了 pandas: pip install pandas
    copy_cluster_images(tasks)
```
